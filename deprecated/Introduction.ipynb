## Purpose of the Guide

Energy practitioners increasingly need reliable climate data to understand how weather and climate conditions influence electricity systems and how these conditions may evolve over time. This guide is thus intended to support the use of climate information in energy sector analysis and decision-making by providing a clear entry point to relevant climate datasets and to illustrate how these data can inform energy-related analyses.The datasets highlighted in this guide are drawn from authoritative scientific sources and represent widely used, state-of-the-art climate information available for research and applied analysis.

## Using Climate Information in the Energy Sector

Energy systems are closely connected to climate conditions, making climate information an important input to many analytical and operational processes. Climate variability and long-term changes can affect electricity demand, resource availability, infrastructure performance, and system reliability. Understanding these relationships helps organizations assess potential risks and consider how future climate conditions may influence energy systems. Climate data therefore provide an important foundation for evaluating future conditions and supporting resilient energy planning.

## How to Use This Guide

The guide is designed as a practical resource that helps users connect energy sector questions with appropriate climate data. It combines explanations of climate datasets with examples of how climate information can be applied in energy sector contexts. Users may approach the guide from different starting points: some may begin with a particular application and seek suitable climate information, while others may start by exploring available datasets. Additional resources help users interpret climate data and understand their appropriate use.

## Supporting Climate-Informed Energy Decisions

By connecting climate datasets with real-world applications, the guide aims to make climate information more accessible and usable for energy practitioners. It provides a structured overview of available data and practical guidance on how climate information can be incorporated into analysis workflows. In doing so, the guide supports more consistent and informed consideration of climate conditions in energy sector planning, operations, and management. The guide also includes a tool for self-assessment, a maturity matrix, that helps organizations reflect on how climate information is currently incorporated into their processes and identify opportunities to strengthen climate-informed practices over time.
