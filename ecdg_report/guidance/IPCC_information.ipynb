## Summary of confidence in direction of projected change in hazard in North America

The latest IPCC report analyses the latest evidence for the direction of the projected change for various hazards. This evidence is summarized in WG1 Table 12.8 ([REF]{.red-text}). This table is reproduced and adapted here. It shows the projected changes in key climatic impact-drivers across North America under mid-century warming conditions (approximately 2°C to 2.4°C global warming). It synthesizes information from the IPCC AR6 WG1 assessment by combining the direction and confidence of projected changes, the expected time of emergence of the climate signal relative to natural variability. 

![Summary of confidence in direction of projected change in climatic impact-drivers in North America, representing their aggregate characteristic changes for mid-century for scenarios RCP4.5 and SSP2-4.5, or above within each AR6 region, approximately corresponding to global warming levels between 2°C and 2.4°C. The table also includes the assessment of observed or projected time-of-emergence of the climate impact-drivers change signal from the natural interannual variability if found with at least medium confidence. </br> (Source: [REF]{.red-text})](image/Table12-8_withmap.png)
