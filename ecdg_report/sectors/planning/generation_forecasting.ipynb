# Generation Forecasting

## Summary

::: {#id1 .callout collapse="true"}

### A detail description...
Electricity generation forecasting under climate change is essential for maintaining a sustainable and reliable power supply. It is required to incorporate climate information in electricity generation and capacity planning. However, a gap remains between advancements in climate data products, models, and analytical methods, and their application in industry practices for planning and implementing generation forecasts. This section bridges that gap by presenting tools and techniques used in both academic research and industry. The first sub-section outlines the climate data products and variables used, along with the methods applied for their analysis, such as downscaling and bias correction, before integrating them into generation forecasting models. Forecasting is conducted using predictive models or methods tailored to specific generation types. Additionally, this section summarizes the current status of electricity generation entities in Canada, based primarily on insights gathered during workshops with industry stakeholders. Finally, the section also examines projected changes in electricity generation for future periods under various emission scenarios, considering both Canadian and regional perspectives.

### Climate data processing methods

The analysis of climate change (CC) impacts on electricity generation systems (EGSs) is not straightforward. It follows several steps (see **Figure 1**), including the acquisition, emission scenarios selection, preparation of appropriate climate data, and model investigations for CC impact analysis on the EGS. Climate data can be collected from many sources, including the most common platform, the Coupled Model Intercomparison Project (CMIP) (@plaga_methods_2023). The CMIP has different release versions, including CMIP3, CMIP5, and CMIP6. The climate projections vary according to the greenhouse gas (GHG) emission scenario, which depicts the influence of anthropogenic climate change. The most common scenarios used to investigate CC impacts on EGSs are A2 (most pessimistic), A1B (moderate), and B1 (most optimistic) from CMIP3, RCP2.6 (low emission), RCP4.5 (medium emission), and RCP8.5 (high emission) from CMIP5, and SSP1-2.6 (low emission), SSP2-4.5 (moderate), SSP3-7.0 (high emission with limited economic growth), and SSP5-8.5 (high emission with strong economic growth) from CMIP6. Depending on the purpose of the analysis, various scenario projections are used for evaluating climate change. **Table 1** presents a summary of climate data, their downscaling and bias correction methods, and the methods utilized for CC impacts on different types of EGSs.

Climate data are obtained from global circulation models (GCMs). These models provide climate projections with a coarse spatial resolution (e.g., 100 km x 100 km), which is impractical for CC impact analysis on EGSs at the regional to local scale. Therefore, it is required to downscale the GCMs' projections to at least the regional scale. Statistical and dynamic downscaling are the two most common methods for spatially downscaling the projections of GCMs. In most cases, statistical downscaling methods are applied to obtain desired finer-resolution climate variables. These methods are based on statistical or empirical relationships and require less computational effort than dynamic methods (@plaga_methods_2023). In recent years, machine learning methods (e.g., RFQ and QRF) have been utilized for statistical downscaling of hydro-meteorological variables from CMIP6 (@zhang_assessment_2022). On the other hand, dynamic downscaling has a complex design structure and a huge computational demand. This type of downscaling is embedded in regional climate models (RCMs) that utilize GCM outputs and lateral boundary conditions to generate regional-scale climate variable projections (@zhang_assessment_2022). 
Some climate data providers have established methods that can simultaneously bias correct and spatially disaggregate climate data from GCMs. For example, the Bureau of Reclamation  has developed a monthly bias-correction and spatial disaggregation (BCSD) method  to downscale and bias correct the CMIP5-GCMs data (@turner_climate_2017).  Some studies (see **Table 1**) have employed standalone bias correction methods that compare climate projections with observations in the reference period and generate coefficients to correct future predictions. This type of statistical method assumes stationarity; however, climate variable projections are nonstationary, and that could add additional uncertainty (@plaga_methods_2023).  A good comparison between statistical and dynamic downscaling methods in terms of their strength and weaknesses is provided in the study of @wilby_sdsm_2002.

### Electricity generation prediction methods
As shown in **Figure 1**, the prediction of electricity generation varies depending on the system types and the variables incorporated into the prediction models. **Table 2** summarizes the models and methodologies employed in previous studies to assess the impact of climate change on electricity generation systems. The following section outlines the approach used to project electricity generation under future climate scenarios for different energy types.

#### Hydropower
In general, downscaled and bias-corrected climate variables (e.g., precipitation and temperatures) from different scenarios are forced into hydrologic models to investigate the hydrologic regime and to generate future flow conditions. These flows are then used as input to hydropower potential estimation models. The distribution of the estimated electricity is compared with the historical observation to analyze the impact of climate change on generation.  In literature, a quite good range of hydropower potential station models or methods, including empirical formula (@parkinson_robust_2015), reservoir optimization algorithm-Sampling Stochastic Dynamic Programming (SSDP) (@haguma_optimal_2014), WEAP (@boehlert_climate_2016) and MODSIM (@kim_climate_2022), have been used. One representative example of an empirical method is represented by **equation 1**.
$$
E=ρghQ\tag{1}
$$
In **equation 1**, $E$ is the hydropower potential with projected streamflow $Q$ at a hydraulic head $h$. $ρ$ and $g$ are the density of water and gravitational acceleration.

Reservoir optimization algorithms (ROAs) are used to determine operation rules or policies that include an average release from hydropower dams and energy production. The projected inflow simulations from the hydrologic model are fed to the ROA to determine operating policies under future climatic conditions obtained from different emission scenarios. Then they are compared with the baseline period policy in order to determine the impact of climate change on hydropower generation. 

Unlike the ROA, MODSIM, a complex surface water network system, can also be used to estimate hydropower generation. This model has an internal optimization operator that can allocate water based on physical and operational constraints. The inflow projection from hydrologic models is used as forcing for this MODSIM model to determine the hydropower generation potential for the future period and compared with the historical baseline period to identify the changes due to climate variability. 

#### Wind Energy
The impact of climate change on future wind energy generation is generally estimated based on the operational wind speed at hub height. Under different emission scenarios (e.g., RCP8.5), future projections for wind speed at 10 m height from surface level are obtained from climate models (e.g., GCMs and RCMs) (@yao_climate_2012, @zhang_assessment_2022, @li_impact_2023). For wind turbine design, cut-in and cut-out wind speeds at hub height play a crucial role, and they typically range from 4 m/s to 25 m/s. Based on the design criteria, hub height varies from 50 m to 120 m, however, the average hub height in Canada is 83 m (@cwtd_canadian_2022). The wind speed at hub height is estimated using the following power-law relation (Equation 2) (@yao_climate_2012, @zhang_assessment_2022, @plaga_methods_2023, @li_impact_2023).
$$
\frac{V_H}{V_{10}} =(\frac{Z_H}{Z_{10}})^α\tag{2}
$$

In **equation 2**, $V_H$ is the wind speed at hub height H ($Z_H$), and $V_{10}$ is the wind speed at 10 m height ($Z_{10}$). 
The exponent $α$ varies with the topographic terrain conditions, codes, and standards. For a natural, stable terrain profile, the $α$ value is 0.14 (@yao_climate_2012, @plaga_methods_2023, @li_impact_2023). The wind speed at hub height is then converted to wind energy by multiplying it by the wind energy density (@li_impact_2023).
$$
W_E = \frac{1}{2} \rho A V_H^3 C_p \tag{3}
$$

In **equation 3**, $W_E$ is the wind energy, ρ is the air density at hub height that varies with the temperature and wind profile, A is the cross-sectional area, and $C_p$ is the power coefficient. For climate change impact analysis, $W_E$ is simulated at each grid point or station, and then trend analysis is conducted. The t-test or the Mann-Kendell approach can be used for trend detection.  In their study, Li (@li_impact_2023) utilized a t-test using a linear regression for trend detection. Another study (@yao_climate_2012) used the Weibull distribution for comparing wind production in the future period under different scenarios with reference to the period.  

#### Solar Energy
Generally, photovoltaic (PV) solar cells are used for electricity (energy) generation. The generation capacity of PV can be expressed with the empirical formula (**equation 4**), consisting of solar radiation and temperature as independent variables (@zhang_assessment_2022). Therefore, two key climate variables, such as surface shortwave radiation (SSWR) and near-surface air temperature (NSAT), are used for the impact analysis of solar energy. The radiation impinging on each solar station is predicted to use SSWR, and the solar station’s ambient temperature is predicted from NSAT. Using historical data of these two variables, solar energy generation curves are produced by applying **equations 4** and **5**. Similar curves are developed for future projections and compared with the reference one developed with historical period data for impact analysis due to climate change.  
$$
C_F=(\frac{S_1}{S_2} )[1+λ_T (T_1-T_2)]\tag{4}
$$
$$
S_E=I_C.C_F.∆_τ\tag{5}
$$
In **equations 4** and **5**, $C_F$ is the hourly capacity factor of the solar station, $S_1$ and $S_2$ are the solar radiation at the location of the solar station and under standard test conditions (ideal 1000 $W/m^2$), respectively. $T_1$ and $T_2$ are the temperatures at the location of the solar station and under standard test conditions, respectively, and $λ_T$ is the temperature coefficient (ideally negative). $S_E$ represents solar energy, $I_C$ is the installed capacity of the station, and $Δτ$ is the hours of period $τ$.

#### Thermal Energy (Nuclear, Natural Gas, Coal, and Biofuel)
Thermal energy encompasses the production of electricity by converting heat into power using nuclear, natural gas, coal, and biofuel sources. The impact of climate change on thermal energy production depends on the cooling systems used by each generation type. Typically, water and air flow (dry cooling) are used to cool down the generator. Therefore, water availability, water temperature, humidity, air pressure, and air temperature play a key role in the thermal power plant's efficiency, which ultimately affects energy generation (@lagace_assessing_2021). Usually, two types of investigation are conducted to determine the impact of climate change on thermal energy systems: 1) regression analysis assuming linear dependency between air and water temperature, and 2) hydrologic modeling to calculate available water for cooling systems (@plaga_methods_2023). 

In a study, Lagace (@lagace_assessing_2021) applied regression models to estimate climate impacts on thermal energy systems in Ontario. The models estimate the derated capacity of thermal systems, including those with recirculating cooling, dry cooling, once-through cooling, and combustion.  These models are based on the air temperature, wet bulb air temperature, relative humidity, and air pressure, and were proposed by Bartos et al. (@bartos_impacts_2016), Henry and Pratson (@henry_differentiating_2019), and Craig et al. (@craig_compounding_2020). 
Like hydropower generation, hydrologic models are forced with future climate scenarios to simulate water flow. Then, the water management model (WMM) is applied to estimate the availability of cooling water under changing climatic conditions. To apply the WMM, a linear relationship is assumed between the available water and the water requirement for the plant’s cooling (@plaga_methods_2023).

### Current status of electricity generation entities regarding climate data, methods and models uses
This section is written based on the information collected during a workshop, and the contents summarize responses from experts representing HydroOne, Manitoba Hydro, Hydro-Québec, IESO, and OPG. A response summary table is provided below (see **Table 3**). For hydrologic model (e.g., WATFLOOD) calibration, several entities rely on bias-corrected reanalysis datasets, which requires corrected precipitation and temperature data. Validation of these datasets is typically performed against observations from Environment and Climate Change Canada (ECCC) stations, Hydro-Québec internal networks, or the Water Survey of Canada. However, some internally interpolated datasets similar to NRCANMET have been noted to produce less reliable results. For future climate projections, GCM/RCM ensembles are commonly used along with multiple scenarios—conservative, moderate, and low—to capture a range of possible outcomes, with data sources including PAVICS (Ouranos), Portraits Climate data, and the Climate Atlas. Regarding suitability for electric generation, no single dataset or model is explicitly identified as optimal; however, short-term forecasts (up to 4 hours), hourly zonal climate variables, and bias-corrected wind speed and solar irradiance are emphasized for wind and solar forecasting. Care is taken to avoid overestimating renewable supply, and model compatibility across different climate data sources is verified to ensure reliable integration into operational planning.

#### Summary
Some electricity generation entities do not use climate change information at all or do not consider potential future climate impacts. Others integrate climate data to quantify projected changes in inflows to generating stations over the next 100 years. In some instances, climate information is applied in resource assessment (e.g., water availability) and plant design, including 100-year flood considerations for new hydro projects.

Climate data are used for risk management and to inform investment planning decisions. Hydrological models (e.g., WATERFLOOD, partial physically based) simulate watershed behavior and forecast inflows using temperature, precipitation, or observed flow data. Statistical approaches (e.g., regression models, runoff factors, fitted distributions) are applied to historical precipitation and streamflow records to estimate floods at different exceedance probabilities. Climate data also support demand and supply forecasting (wind, solar, transmission/transfer limits) through statistical models, while physics-based models such as PSS/E are used for load flow simulations. Overall, entities use physical, statistical, and probabilistic models, often combining them with climate information.

Electricity generation entities rely on temperature data (hourly, daily min/max, seasonal averages), wind speed and gusts, solar irradiance, humidity, and cloud cover, often by specific zones. Precipitation data include rainfall, wet snow, freezing rain, snowpack, and extreme storm events (e.g., 100-year floods), with evaporation, lake water temperature, and atmospheric moisture also considered for hydropower planning. These data feed into generation profiles for wind and solar, hydro inflow forecasts, and hydraulic models, alongside watershed-scale precipitation and river flow patterns. Emphasis is placed on extreme events and long-term climate trends, averages, and frequencies to assess design conditions and operational risks. Currently, most profiles are based on historical observations, though some entities are exploring uncertainty analyses under climate change scenarios.

Entities require a mix of historical data, reanalysis products, seasonal forecasts, and long-term climate projections to support design, planning, and operations. Historical inflows, wind, and solar inputs remain the primary basis for current designs, with some adjustments using climate sensitivities. Reanalysis and seasonal forecasts are valuable for weekly to seasonal streamflow, snow level, and hydrologic simulations. Forecasted solar irradiance and wind patterns are used to estimate renewable generation potential. Real-time, seasonal, and long-term forecasts together enable better operational decision-making and strategic planning under changing climate conditions.

Climate data needs range from province-wide coverage to highly localized site-specific observations, depending on the project and application. For hydrologic modeling, watershed-scale temperature and precipitation data are essential, with some applications requiring gridded datasets at resolutions as fine as 1-2 km². Regional analyses may involve multiple locations, proximity to transmission systems, and zonal breakdowns (e.g., 7 zones in Quebec, IESO zones in Ontario). Current practices often use point data near population centers or the closest grid cell to generation stations for variables such as wind speed, temperature, cloud cover, and solar irradiance. For single-station frequency analyses, data from one location is sufficient, while broader hydrometeorological and operational planning requires multi-location and watershed-wide coverage.

Entities require climate data at hourly or sub-hourly resolution, with daily values also useful for certain applications, and records ideally spanning at least the past 40 years. Horizontal resolution typically ranges from 2 km to 10 km grids, with some utilities using 6 km x 6 km or 10 km x 10 km datasets for temperature and precipitation. Seasonal data help identify trends and potential risk areas, while short-term (up to 4-hour) forecasts support operational needs. Vertical resolution requirements include ground-level measurements for temperature and solar, and wind speed data at multiple elevations from 10 m to hub height (up to ~100 m). This combination of fine spatial detail, high temporal resolution, and vertical profiling enables accurate modeling for hydro, wind, and solar generation.

Entities typically require at least 30–40 years of historical climate data, with some using records going back to the 1950s to establish robust baselines. For planning and adaptation, future climate projections are needed over horizons of 30–100 years, often extending to 2100 to match asset lifespans. Forecast horizons range from 15 to 30 years for medium-term planning, with operational forecasts extending up to 14 days ahead. Some applications require continuous, year-round real-time data, often at hourly resolution, alongside long-term datasets. This combination of historical records, near-term forecasts, and century-scale projections supports both day-to-day operations and strategic infrastructure planning.

Update needs vary by application, with long-term climate projections often refreshed every 2–5 years to align with CSA standards and bulk planning cycles, and some datasets, such as IDF curves, are updated every 5–10 years. Certain extreme event parameters (e.g., PMP, PMF) may only be revised every 10–25 years. Real-time operational data is expected with latencies under 30 minutes, with short-term forecasts updated daily or every 6 hours. Annual or periodic updates help incorporate advances in climate science and maintain realistic future projections. In some cases, specific variables like temperature, ice, or wind are updated on longer 10–20-year cycles depending on their stability and monitoring requirements.

Entities rely on a combination of national and provincial meteorological and hydrological datasets, including ECCC climate stations, Water Survey of Canada, Hydro-Québec networks, and internal operational monitoring networks for streamflow, snow, and wind. Reanalysis and climate projection products are sourced from CASR, CaPA, PAVICS, ECWMF, Portraits Climate data, Climate Atlas, IREQ, and scientific literature. Additional data comes from consultants (e.g., UL for GHI and windspeed), NOAA, NRCAN, StatsCan, and paid providers, as well as provincial resources like Ontario Flood Maps, MNRF Fire Maps, and Hydro-Québec’s SIGAL platform. Grid emissions, temperature, and precipitation data are used for planning and design, often integrated with historical records or near-real-time observations. These diverse sources support hydro, wind, solar, and building code assessments, as well as long-term climate change analyses.

Many data products rely heavily on historical records, which may not capture future climate changes or current baseline conditions. Limitations include short records for flow, ice, and rating curves, measurement biases (e.g., precipitation undercatch, wind exposure), and uncertainties in reanalysis or climate projections. Some datasets, like ERA5, may have known biases, and reanalysis products are not always suitable for trend analysis. To address uncertainty, ensembles and multiple future scenarios are often used, but the understanding of limitations remains incomplete. Internal and external datasets are partially relied upon, with calibration and alignment to standards used to mitigate known constraints.

Frequent, long-term updates and easy access to climate data that can be integrated into internal models and maps remain limited. High-resolution datasets, including humidity, temperature, wind, precipitation, and solar irradiance at granular spatial and temporal scales, are often unavailable, particularly for rare extreme events or compound conditions like ice accumulation with wind or snowmelt with rain. Historical and future projections for long horizons (50–70+ years) are scarce, and verification of forecasted data is limited. Methodological frameworks, computational tools, and internal expertise to interpret and apply climate variables for operational or planning decisions are often lacking. Gaps also exist in linking climate data to renewable energy production, assessing extremes (e.g., tornadoes, forest fires, droughts), and achieving adequate spatial and temporal coverage.

Entities require the ability to downscale GCM/RCM outputs and assess streamflow forecast skill using metrics such as Nash-Sutcliffe Efficiency, KGE, CRPS, and Brier scores, particularly for seasonal ensembles. Rainfall and hydrologic data must have sufficient temporal resolution to evaluate short-duration events, with inflows used to validate hydrologic models. For extreme events like PMP, formal validation is limited, relying instead on expert judgment and professional assessment. Data quality checks include identifying outliers, consistent biases, and general reasonableness, often supported by consultant review and high-level QA/QC. Overall, climate data is primarily used for long-term planning, leveraging available and accessible datasets while balancing formal metrics with knowledge-based evaluation.

Validation of climate data involves comparing reanalysis, model outputs, and historical datasets against observed measurements to assess accuracy and detect biases. Hydrologic and energy models are calibrated using reference data, though internally interpolated or boundary-condition datasets may not always produce expected results. Cross-verification with observed wind, solar irradiance, and streamflow helps benchmark forecasts and identify systematic errors or drifts over time. Regular calibration ensures consistency in data formatting, time stamps, coordinates, and alignment with geographic boundaries. Multiple climate change models are used to evaluate future extremes, annual average flows, and seasonal shifts from dry to wet periods.

Uncertainty is quantified using ensembles, confidence intervals, percentiles, standard deviations, and visualizations to represent possible variations in climate variables. Raw ensemble outputs, often bias-corrected, are directly fed into hydrologic, wind, and solar models to evaluate operational and infrastructure risks. Multiple climate models and scenarios (conservative, moderate, and low) are used to capture a range of possible outcomes for design and planning purposes. Knowledge of uncertainty helps assess coverage of design criteria, inform decision-making, and guide risk management. Curve fitting and statistical analyses are applied to interpret data variability and support scenario evaluation.

Data biases arise from measurement errors, equipment changes, station relocations, land cover effects, and limitations in precipitation, wind, or ice observations. Reanalysis products and future climate models each carry their own biases, including wet biases in precipitation or diurnal wind pattern errors. Such biases can affect infrastructure design, component reliability, and the accuracy of short- and long-term forecasts, potentially leading to over- or under sizing of assets. Correction and downscaling techniques are applied where possible, particularly for wind speed and precipitation, while biases in wind direction are less critical. Despite corrections, residual biases influence projections and forecasts, necessitating calibration and careful interpretation using delta or ensemble analyses.

Biases and low-quality data can propagate through models, affecting planning, operations, and reliability across multiple departments. Impacts are generally asymmetric: underestimation can lead to insufficient infrastructure, unmet service levels, or inadequate energy reserves, while overestimation can increase costs but often has less critical reliability consequences. Poor or biased climate data can compromise peak demand predictions, renewable generation scheduling, and system response to droughts or extreme events. Bias corrections, calibration, and ensemble approaches are applied to mitigate these impacts, but residual errors can still lead to suboptimal maintenance, market decisions, or regulatory non-compliance. Overall, the consequences of inaccurate data are context-dependent and can result in inappropriate decisions if not carefully managed.

Models are generally susceptible to biases in climate data, with small errors potentially producing unreliable forecasts or operational outputs. Hydrologic models like WATFLOOD and weather or energy forecasting models are susceptible to inflows, load, and temperature variations. Biases in key variables, such as temperature or wind/solar supply, can significantly affect thermal capacities, energy scheduling, and generation estimates. Recalibration with bias-corrected datasets and verification across different models is essential to mitigate sensitivity. While some models incorporate bias adjustments, the exact magnitude of sensitivity often remains uncertain and context dependent.

### Impacts of Climate Change on Electricity Generation Systems (EGSs)
#### Hydropower 
The three hydropower studies  (see **Table 4**) used different climate model datasets and projection periods to assess the impacts of climate change. 

Manitoba: @kim_climate_2022 applied CMIP5 projections from 14 GCMs under RCP4.5 and RCP8.5 for the Lower Nelson River Basin, Manitoba, comparing a historical period (1981–2010) with mid-century (2021–2050) and late-century (2041–2070) scenarios. Hydrologic models were used to simulate hydrology and local flow for input to MODSIM-DSS operatioon model for reservoir simulation and hydropower prediction (**Table 2**). They found substantial seasonal warming, with the largest temperature increases in spring and winter, alongside higher annual and seasonal precipitation and significant winter runoff gains but summer reductions. These changes translated to increased annual hydropower potential but notable seasonal losses in winter and summer/fall. 

Quebec: @haguma_optimal_2014 used CMIP3 projections from 13 GCMs under A1B, A2, and B1 scenarios for the Manicouagan River Basin, Quebec, comparing 1970–1999 with three future periods (2010–2039, 2040–2069, 2070–2099). Hydrologic model (SWAT) was used to generate reservoir inflow prediction and Reservoir optimization algorithm (SSDP) was utilized to determine future operating rules under climate change (**Table 2**). Results indicated pronounced winter warming and precipitation increases, earlier spring peak flows, and higher annual generation potential, but also operational inefficiencies from unproductive spills and reduced generation efficiency. 

Brithish Columnbia: @parkinson_robust_2015 used PCIC downscaled outputs from 8 GCMs under B1, A1B, and A2 scenarios for British Columbia, comparing 1961–1990 with 2041–2070 projections. They applied hydrologic model to estimate future streamflow and empirical electricity model to estimate future hydropower. They reported earlier spring snowmelt, increased winter and spring runoff, reduced summer runoff, and an overall increase in annual generation, with large seasonal shifts favoring winter output at the expense of summer supply. Collectively, the studies highlight a common pattern of seasonal hydrological shifts driven by warming and earlier snowmelt, with increased winter energy production but risks to summer generation reliability.

#### Wind Energy
The two wind energy studies (see **Table 4**) used different climate model datasets and projection periods to assess the impacts of climate change. 

Canada: @li_impact_2023 used CanRCM4 simulations from 15 RCMs under the RCP8.5 scenario, with 0.4° spatial resolution and hourly data, to assess wind energy changes across North America. The analysis compared a historical baseline (1986–2006) with mid-century (2050–2070) and late-century (2080–2100) periods. Empirical formula (power-law relation) for wind speed estimation at hub height was utilized.  Results show a 5–10% increase in 20-year return period wind speeds in western coastal British Columbia, northern Quebec, and eastern coastal regions by 2070, with further northern increases by 2100. Wind energy potential (WE) changes are modest in southern Canada (<5% by 2070; <10–15% by 2100) but larger in the western coastal, far northern, and Hudson Bay regions (+10% by 2070; +15–20% by 2100). Seasonally, most of Canada shows a positive WE trend in spring but a negative one in summer, with mixed fall responses. The Hudson Bay and certain coastal areas may experience increases exceeding 20%, while some east and west coastal zones may see decreases of 5–10%.

Ontario: @yao_climate_2012 used the PRECIS regional model (HadRM3P) driven by the HadAM3H GCM at 50×25 km resolution under A2 and B2 scenarios, combined with Canada Wind Energy Atlas reanalysis data, to assess changes for Ontario between 1961–1990 and 2071–2100. They used empirical formula (power-law relation) for converting wind speed to wind. They projected small declines in wind speed (1–3%) in southern Ontario, but notable increases in northern water bodies such as James Bay (+10% under A2). WE is projected to decrease significantly in southern Ontario (−14% under A2; −9.3% under B2), while offshore wind power density in James Bay and Georgian Bay could increase substantially (+33.1% and +9% under A2; +13.1% and +14.4% under B2). These results indicate a regional redistribution of wind energy potential, with southern areas losing capacity and northern/offshore zones gaining.
