## Role in the Electricity System

Dam main purpose is to store water or other liquid for energy production among other uses such as water irrigation. However, dams can pose significant risks to public safety and the environment if they are not maintened and operated properly. Therefore, dam safety is a critical aspect of infrastructure management in the electricity sector as many hazards can cause increasing risks on these structures .

"A Dam Safety Review involves a systematic examination and assessment of the data and information about the design, construction, maintenance, operation, processes, and systems affecting a Dam’s safety, including the Dam safety management system. " (@damsafetybc) Overall, a dam safety review is a systematic evaluation of all aspects of a dam based on currrent knowledge and standards (@damsafetyontario) which may not be the same as the ones used in the dam construction. The review needs to be carried out by profesionnal engineers. 



Each province and territory in Canada has its own regulations and guidelines for dam safety, which are informed by the CDA's Dam Safety Guidelines. Most provinces and territories provide their own guidance to support the implementation of these guidelines, and some have additional requirements or standards that must be followed (see, https://cda.ca/dams-in-canada/regulation).
