## Summary Description

The [CO]{.underline}ordinated [R]{.underline}egional [D]{.underline}ownscaling [EX]{.underline}periment ([CORDEX](https://cordex.org)) is a framework of the World Climate Research Program ([WRCP](https://www.wcrp-climate.org/)) aimed at the production and evaluation of [regional climate model](Climate_Projection_Datasets.ipynb#Regional Climate Models) projections. CORDEX is based on the vision of advancing and coordinating the science and application of regional climate downscaling. Through global partnerships it works to achieve its goals to better understand regional and local climate phenomena, improve and evaluate regional downscaling models and techniques, produce coordinated sets of regional downscaled projections, and to promote communication and knowlege exchange with users of regional climate information.
The regional projections are produded over 14 domains covering most of the Earth's major land masses (see figure below).
The CORDEX ensemble providing downscaled CMIP6 global climate model simulations for the North American domain provides a subset of the CMIP6 models simulations, and continues to grow as teams contribute their RCM simulations. The previous generation of CORDEX downscaled CMIP5 simulations from 7 GCMs using 9 differen RCMs, filling a [sparse matrix of simulations](https://na-cordex.org/simulation-matrix.html) at 3 different spatial resolution and using 3 differnt RCPs.

## Dataset Characteristics

- **Current version:** CORDEX-CMIP6  
- **Temporal coverage:** 1950–2100  
- **Temporal resolution:** from hourly to annually, all models provide daily. Dependent on model and CORDEX generation.   
- **Spatial coverage:** 14 continental domains  
- **Spatial resolution:** 0.11° (\~12 km), 0.22° (\~25 km), 0.44° (\~50 km); availability depends on models and CORDEX generation.  
- **Data type:** raw model outputs and bias-adjuste climate projections

## Strengths and Limitations

### Key Strengths of CORDEX

| Strength | Description |
|------|-------------|
| **High Spatial Resolution** | RCMs resolve regional and local climate patterns better than GCMs, particularly in regions of small scale surface features such as mountain terrain and coastal areas. |
| **Comparable Simulations** | CORDEX provides the framework to produce and provide comparable simulations following a shared producion protocol. |
| **Multi-Model Ensembles** | Provides simulations from multiple regional climate models (RCMs)  driven by multiple GCMs. |

: {.tbl .striped .light .sm}

### Key Limitations of CORDEX

| Limitation | Description |
|------|-------------|
| **Reduced sampling** | Limited by computational capacities, RCM modeling centers will downscale a subset of GCMs only, thus CORDEX ensembles may not sample the full range of model uncertainty, and emission scenario uncertainty. |
| **Time Lag of Availability** | CORDEX data rely on GCM driving data, thus can only be produced after a new generation of GCM simulations have been made available. Hence, RCM ensembles lag behind the latest available GCM data. |
| **Large Data Volume** | High-resolution and multi-member ensembles require significant storage and processing capacities. |
| **Biases Remain** | Dynamical downscaling may reduce, but does not eliminate, GCM biases. |

: {.tbl .striped .light .sm}

## Expert Guidance

::: {#guidance .callout collapse="true"}
### Click to toggle section collapse
The [CO]{.underline}ordinated [R]{.underline}egional [D]{.underline}ownscaling [EX]{.underline}periment ([CORDEX](https://cordex.org)) ensembles of regional climate model (RCM) simulations provide historical climate simulations and projection data at finer spatial scales for impacts and adaptation work. As a global framework, CORDEX is a diagnostic MIP within CMIP that coordinates domains, protocols, and evaluation practices, emphasizing that RCMs inherit boundary-condition uncertainty from their driving GCMs and that ensemble interpretation must reflect this dependency (@Gutowski2016).

Canada is covered by the North American domain and partially by the Arctic domain. North America CORDEX maintains its own website ([NA-CORDEX](https://na-cordex.org/index.html)) which covers the previous generation of RCM simulations that were driven by CMIP5 global simulations and updates for CMIP6 driven simulations are pending. The CORDEX ensemble is built from the contributions of regional modeling centers who choose the respective downscaling experiments they perform. Thus, not all RCMs will downscale the same, and certainly not all, GCM simulations. When using an RCM ensemble it must be noted that, by definition, it may sample a different uncertainty space compared to the driving CMIP ensemble.

Systematic evaluations of over Europe showed credible continental-scale temperature and precipitation patterns and documented model-specific biases, illustrating the conditional “added value” of higher resolution for orography, coasts, and extremes (@Kotlarski2014). @Lucas-Picher2017 explored dded value of the Canadian Regional Climate Model's (CRCM5) simulations over North America. An evaluation of NA-CORDEX-CMIP5 by @Bukovsky2020 shows that the downscaled NA-CORDEX ensemble reproduces well the CMIP5 global simulations projections and climate sensitivity. This holds for temperatures at local to continental scales, but only for continental scales for precipitation. The spread of the NA-CORDEX-CMIP5 ensemble is similar for temperature but larger for precipitation. A description of the NA-CORDEX data archive is provided by @McGinnis2021, including bias-corrected simulations.

Published application of NA-CORDEX-CMIP5 in the energy sector include a study of climate change impacts on wind ressources (@Chen2020) and a cooling and heating energy demand analysis (@Tian2022). @Ganguli2019 did an assessment of future changes in intensity-duration-frequency curves for Southern Ontario using NA-CORDEX data. Other analysis and research using NA-CORDEX including hydrological studies can be found on the [NA-CORDEX web site's comprehensive bibliography](https://na-cordex.org/bibliography.html).  

Generally, the strengths of CORDEX are (i) finer resolution that can better represent topography, land–sea contrast, mesoscale circulations, and some extremes; (ii) coordinated protocols that enable multi-model, comparisons; and (iii) region-tailored ensemblesand the globally consistent [CORDEX-CORE set](https://cordex.org/experiment-guidelines/cordex-cmip5/cordex-core/cordex-core-simulations/) for broad coverage.

Key limitations include (i) the inheritance of GCM biases and scenarios — RCMs do not “fix” incorrect large-scale circulation; (ii) non-independence and sparse matrices—ensembles often have shared components and incomplete cross-pairings of GCMs and RCMs, so simple equal-weight means can be misleading; (iii) process and parameter uncertainties — convection, clouds, and land–atmosphere coupling remain challenging; and (iv) heterogeneous availability and formatting, especially during the CMIP6 transition, requiring careful selection, bias adjustment, and documentation checks. 

Application of CORDEX should start from the most complete regional matrix available and verify evaluation runs (reanalyses-driven) to understand baseline skill. The selection of driving GCMs should span regional circulation patterns and sensitivities, and insure robustness by using multiple RCMs and multiple realizations where available. GCM selection approaches for CORDEX-CMIP6 have been documented by @DiVirgilio2022 for Australia and by Sobolowski2025 for EURO-CORDEX.

:::

## Available Variables in CORDEX

CORDEX regional models may produce variables from the the [IPCC list of standard output from Coupled Ocean-Atmosphere GCMs](https://pcmdi.llnl.gov/mips/cmip3/variableList.html). It comprises far over 100 different variables, and many more may be produced by a model (see the [Climate and Forecast (CF) Standard Name Table](https://cfconventions.org/Data/cf-standard-names/current/build/cf-standard-name-table.html)). Depending on the model and its realizations, the respective list of available variables will be determined by the modeling center's available resources and research focus. Hence, from some models many atmospheric and surface variables will be available, however the common denominator for most CORDEX model simulations will be the variables listed below.

*Click on variable groups to uncollapse*

:::: {#tas .callout collapse="true"}
### Temperature
Temperature variables are provided at hourly to daily intervals, depending on the model and realization.

- mean temperature (day, 6hr, 3hr, 1hr) [K]
- daily minimum temperature [K]
- daily maximum temperature [K]
:::

::: {#pr .callout collapse="true"}

### Precipitation
Precipitation is commonly provided as the sum of solid and liquid precipitation or as snowfall and variables are provided at hourly to daily intervals, depending on the model and realization. 

- Precipitation flux (day, 6hr, 3hr, 1hr) [kg m^-2^ s^-1^]
- Snowfall flux [kg m^-2^ s^-1^]
:::

::: {#other .callout collapse="true"}

### Other Variables
Other variables often provided by a large number of models include:

- Wind (estward, westward) [m s^-1^]
- Specific or relative humidity [%]
- Longwave and shortwave radiation (downwelling and upwelling) [W m^-2^]

Further variables related to evaporation, runoff, and cloud cover may be available depending on the model.
:::

## Data Access

The web portal for CORDEX North America provides information for [data access](https://na-cordex.org/data-access.html), pointing to  [NCAR's Goescience Data Exchange](https://gdex.ucar.edu/datasets/d316009/) and respective Earth System Grid Federation ([ESGF](https://esgf.github.io/index.html)) [Federated Metagrid Nodes](https://esgf.github.io/nodes.html).
The CMIP6 driven simulations using the Canadian Regional Climate Model (CRCM) [add reference when available]{.mark-red} are available on [PAVICS](https://pavics.ouranos.ca/datasets.html#a) for the North American continent. To avoid downloading very large datasets in their entirety PAVICS allows partial/regional extraction and provides [tutorials](https://pavics.ouranos.ca/climate_analysis.html#a) to do so. With a free PAVICS user account, the Jupyter notebook with the Python code from the tutorials can be [directly used on PAVICS](https://pavics.ouranos.ca/jupyter/hub/user-redirect/lab/tree/pavics-homepage/PAVICStutorial_ClimateDataAnalysis-1DataAccess.ipynb).
