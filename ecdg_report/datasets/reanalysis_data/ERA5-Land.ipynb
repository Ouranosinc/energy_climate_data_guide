# ECMWF ERA5-Land and ERA5

## Summary Description

ERA5-Land is a high-resolution global land surface reanalysis dataset produced by the European Centre for Medium-Range Weather Forecasts ([ECMWF](https://www.ecmwf.int/)). ERA5-Land is a replay of the land component of the [ERA5](https://cds.climate.copernicus.eu/datasets/reanalysis-era5-single-levels) climate reanalysis, forced by meteorological fields from ERA5, but at higher resolution. No data assimilation is used in the production of the dataset, but the observations indirectly influence the simulation through the atmospheric forcing of ERA5. The dataset provides historical climate data of land surface variables at high horizontal (9 km) and temporal resolution (hourly) over an extended time period (1950-present). ERA5-Land was designed to support hydrological studies and diverse surface applications dealing with water resource, land, and environmental management.

## Dataset Characteristics

- **Current version:** Based on ERA5 the dataset is operationally updated daily.
- **Temporal coverage**: 1950 to present
- **Temporal resolution:** Hourly and daily 
- **Spatial coverage:** Global land area
- **Spatial resolution:** ~9 km grid spacing (0.1˚)
- **Data type:** ECMWF land surface model forced by ERA5 atmospheric conditions

## Strengths and Limitations

### Key Strengths of ERA5-Land

|Strength|Description|
|-|---|
|**High spatial resolution**| 0.1° x 0.1°; Native resolution is roughly 9 km x 9 km.| 
|**Long-term consist data**| Extends back to 1950, enabling long-term climate studies and trend analysis (with caution).| 
|**Data timeliness**| The product is updated in near real time on a daily basis. |
|**Hourly data availability**| Supports high-frequency climate and hydrological applications.|  
|**Improved land surface processes**| Enhanced representation of soil moisture, snow, and vegetation dynamics compared to ERA5.|  

: {.tbl .striped .light .sm}

### Limitations of ERA5-Land

|Limitation|Description|
|-|---|
|**Precipitation uncertainties**      |ERA5-Land does not assimilate precipitation observations but interpolates  precipitation which was generated by the ERA5 reanalysis.|  
|**Limited ocean representation**    |Focuses on land, no data is available over oceans. |
|**No assimilation of land observations**     |Many surface variables (e.g. soil moisture, snow depth) are strongly model-driven and only indirectly constrained by observations. This introduces structural uncertainty that can be difficult to quantify.|
|**Wind data only at 10 m height**   |Wind data are only provided at 10m. However, 100 m data aree available from [ERA5](https://cds.climate.copernicus.eu/datasets/reanalysis-era5-single-levels).|
|**Coastal regions poorly resolved**| Due to the coarser resolution of the input dataset ERA5 coastal areas may not be covered by ERA5-Land. For example Les Îles de la Madeleine in the Golf of St Lawrence are not represented.|

: {.tbl .striped .light .sm}

## Expert Guidance

::: {#id .callout collapse="true"}
### Click to toggle section collapse
ERA5-Land is driven by atmospheric variables from ERA5, such as air temperature and humidity, which are used to force the land surface model. Although in-situ observations are not assimilated directly into ERA5-Land, they still influence the product indirectly through the ERA5 atmospheric forcing that constrains the land simulations.

The ERA5-Land dataset provides a consistent long-term view of the evolution of land surface variables, which makes it an interesting product for certain applications. However, one should be aware that the uncertainty of the model estimates is larger in the earlier years of the dataset. This is due to the lower number of observations available for assimilation to create a good quality atmospheric forcing at that time. Compared to its driver ERA5, beyond higher resolution, ERA5-Land shows added value in the description of the hydrological cycle and river discharge, with enhanced soil moisture and lake description. Validation has shown that a good part of the better performance is related to better spatial representation of energy fluxes near coasts or lakes. It is noteworthy to point out that both ERA5 and ERA5-Land outperform their predecessor ERA-Interim, and the multi-decadal consistency enables reliable trend analysis (@Munoz-Sabater2021).

#### Temperature
Temperature values of ERA5-Land are generally well reproduced, however, because cities are non-existent and ERA5-Land does not assimilate observations, the urban heat island signals in cities aren’t explicitly represented. For peak-demand and heat-risk work in cities, urban stations may be a better alternative. For a comparison of the applicability of ERA-Land in the identification of extreme temperature events see @Sheridan2020.

#### Precipitation
Precipitation is generated by the ERA5 atmospheric model at 35km resolution and interpolated to the 9km grid of the ERA5-Land land surface model, which does no adjustment or correction, and no direct rain-gauge assimilation is performed. Therefore, the actual resolution of the precipitation field is much lower than 9 km. ERA5-Land precipitation values generally show large deviations from observations (@Munoz-Sabater2021), partially because observations are point observations integrated over a period of time, rather than representing averages over a model grid box and model time step.

#### Wind
Local terrain, vegetation and buildings may not be represented in the dynamical model used to construct the reanalysis, thus the surface wind can be significantly different from observations.
At ~9 km resolution, complex terrain and shoreline circulation are under-resolved, so local speedups/channeling and lakes/sea-breeze structures are often underestimated. This can be important for BC fjords & mountain valleys and along the Great Lakes (ON/QC). ERA5-Land has been reported to substantially underestimate near surface wind speed at glacier sites in the Rockies (@Draeger2024).  Note that ERA5-Land only provides 10 m winds and it inherits these from ERA5. @Chen2024 show for ERA5 that surface winds in cyclones tend to be weaker than observed at stations. Frequently required wind data at 100 m (approx. hub height) is available from [ERA5](https://cds.climate.copernicus.eu/datasets/reanalysis-era5-single-levels).

#### Snow
Regarding snow accumulation, ERA5-Land shows improved snow mass and snow depth estimates in mid-latitude mountains which is related to the higher resolution. However, ERA5 and ERA5-Land both overestimate total Northern hemisphere SWE by up to 150 % to 200 % compared to the SWE reference data (for details see @Kouki2023 and @Sarpong2025). The discrepancies in SWE are primarily driven by biases in snow depth rather than snow density. Positive trends in SWE tend to be dampened while negative trends tend to be exaggerated.

#### Lakes
It should be noted that the ERA5-Land model does not include the water balance equation for lakes and lake volume will be held constant (i.e., lake depth and surface area do not change in time). Thus evaporation and mixing depth will be based on a fixed, “average” water depth. Generally, lake water surface temperature is underestimated (1.3˚C cooler than observed, on a global average). Hence, ERA5-Land’s lake temperatures and energy fluxes are fine for broad-scale climatology but cannot be used directly for studies needing realistic water level or storage changes (@Munoz-Sabater2021)

#### Permafrost
@Cao2020 concluded that ERA5-Land soil data are not well suited for informing permafrost research and decision making over Northern Canada.

#### Other relevant information
Evaluation studies of ERA5 and ERA5-Land have been conducted for applications in agriculture in Manitoba (@FatolahzadehGheysari2024) and their use in hydrology (@Tarek2020). A study by @Siles2023 compares ERA5-Land ice thickness and lake ice temperature variations to satellite observations for a hydroelectric reservoir in Quebec. @Betts2019 validate ERA5-Land data for temperature, precipitationn, wind, downwelling shortwave and longwave radiation using station data in Saskatchewan.
:::

## Variables available in ERA5-Land

*For details click on variable group to uncollapse*

::: {#id .callout collapse="true"}
### Temperature  

- 2m temperature [K]
- 2m dewpoint temperature [K]

Temperature of air at 2m above the surface of land, sea or in-land waters. Temperature values are provided in kelvin and can be converted to degrees Celsius (°C) by subtracting 273.15.</br>
Dewpoint temperature can be used to calculate relative humidity when combined with temperature and pressure.</br></br>
For more details see the respective [table at the bottom of the ERA5-Land dataset web site](https://cds.climate.copernicus.eu/datasets/reanalysis-era5-land?tab=overview).
:::

::: {#id .callout collapse="true"}
### Precipitation

- Total precipitation [m]

Precipitation values comprise accumulated liquid and frozen water, including rain and snow, originating from both, large-scale precipitation and convective precipitation.

For more details see the respective [table at the bottom of the ERA5-Land dataset web site](https://cds.climate.copernicus.eu/datasets/reanalysis-era5-land?tab=overview).
:::

::: {#id .callout collapse="true"}
### Wind

- 10m u-component of wind (eastward) [m s^-1^]
- 10m v-component of wind (northward) [m s^-1^]

Wind speed is provided as eastward and northward components of the 10m wind, describing the horizontal speed of air moving at a height of ten meters above the surface. The northward and eastward components can be combined to give the speed and direction of the horizontal 10m wind.
</br></br>
For more details see the respective [table at the bottom of the ERA5-Land dataset web site](https://cds.climate.copernicus.eu/datasets/reanalysis-era5-land?tab=overview).</br></br>
Note that frequently required hub-height 100 m wind data at are available from ERA5-Land's driving dataset [ERA5](https://cds.climate.copernicus.eu/datasets/reanalysis-era5-single-levels).
:::

::: {#id .callout collapse="true"}
### Humidity

ERA5-Land does not provide any variable for humidity. However, relative humidity may be calculated from temperature, dew point temperature and surface pressure. For more details see the respective [table at the bottom of the ERA5-Land dataset web site](https://cds.climate.copernicus.eu/datasets/reanalysis-era5-land?tab=overview). Relative humidity can be calculated with respect to water or ice. Water is usually used for observations and is recommended by the WMO. This is most likely what users will be familiar with. Ice is usually used in models.
:::

::: {#id .callout collapse="true"}
### Radiation  

- Surface net solar radiation [J m^-2^]
- Surface net thermal radiation [J m^-2^]
- Forecast albedo [fraction]
- Snow albedo [fraction]

Solar radiation (shortwave) includes both direct and indirect incoming radiation. Solar and thermal radiation are accumulated energy from the beginning to the end of the time step. By model convention downward fluxes are positive. To obtain values to [W m^-2^] the accumulated values need to be divided by the accumulation period expressed in seconds.
:::

::: {#id .callout collapse="true"}
### Snow  

- Snow cover [%]
- Snow density [kg m^-3^]
- Snow depth [m]
- Snow depth water equivalent [m of water equivalent]
- Snow evaporation [m of water equivalent]
- Snowfall [m of water equivalent]
- Snowmelt [m of water equivalent]

Most values for snow are gid box averages, exept the snow depth water equivalent which refers to the snow covered fraction of the grid box.</br></br>
For more details see the respective [table at the bottom of the ERA5-Land dataset web site](https://cds.climate.copernicus.eu/datasets/reanalysis-era5-land?tab=overview).
:::

::: {#id .callout collapse="true"}
### Other Meteorological Variables  

ERA5-Land provides a host of other surface and meteorologial variables such as potential evapotranspiration, surface pressure, surface sensible heat flux, snow layer temperature, soil water content and temperatures, lake water parameters, leaf area index, and runoff variables.</br></br>
For more details see the respective [table at the bottom of the ERA5-Land dataset web site](https://cds.climate.copernicus.eu/datasets/reanalysis-era5-land?tab=overview).
:::

## Data Access

ERA5-Land hourly data from 1950 to present are available on the [Copernicus Data Store](https://cds.climate.copernicus.eu/datasets/reanalysis-era5-land?tab=download). ECMWF has a web site with [instructions to download ERA5-Land](https://confluence.ecmwf.int/display/CKB/How+to+download+ERA5).
Hourly and daily temperature and precipitation data from ERA5-Land over North America are available on [PAVICS](https://pavics.ouranos.ca/datasets.html#c).

To avoid downloading very large datasets in their entirety PAVICS allows partial/regional extraction and provides a [tutorial](https://pavics-sdi.readthedocs.io/en/latest/notebooks/CaSR_basic.html) to do so, using the CaSR reanalysis as an example . With a free PAVICS user account, the Jupyter notebook with the Python code in the tutorial can be [directly used on PAVICS](https://pavics.ouranos.ca/jupyter/hub/user-redirect/lab/tree/tutorial-notebooks/CaSR_basic.ipynb).

ERA5 data can be obtained from the [Copernicus web site](https://cds.climate.copernicus.eu/datasets/reanalysis-era5-single-levels?tab=download).
